# Week 1: an image as an array

**What we do.** We load test images, look at how a computer stores them (an array of numbers), convert a colour image to the YCbCr colour space and display images with labelled axes.

**What to expect.** A greyscale image is a 2D array of shape `(rows, cols)`; a colour image is a 3D array `(rows, cols, 3)`. In this course all images are `float64` arrays with intensities between 0 (black) and 255 (white). The luminance channel `Y` of YCbCr looks like a greyscale photograph, while `Cb` and `Cr` carry only colour.

**How to read the code.** Every function we call comes from the course package `wavelets_lab` (folder `src/wavelets_lab/`); the notebook itself defines no functions. The cell below imports Matplotlib (`plt`, plotting), NumPy (`np`, arrays) and the module `images`. The last section of the notebook describes every function used here in detail, with instructions for writing it yourself.

````python
import matplotlib.pyplot as plt
import numpy as np

from wavelets_lab import images
````

## Loading images

**What we do.** We load two licence-free test images from scikit-image: `camera` (a greyscale photograph) and `astronaut` (a colour photograph), and inspect their shape, type and range of values.

**What to expect.** `camera` has shape `(512, 512)`, `astronaut` has shape `(512, 512, 3)`; both are `float64` and their values lie in `[0, 255]`.

**How to read the code.** `images.load_test_image(name)` returns the image as an array. The last line of a cell is displayed automatically by Jupyter; here it is a tuple of five values: the shape of `camera`, its data type, its smallest and largest value, and the shape of `astronaut`.

````python
camera = images.load_test_image("camera")
astronaut = images.load_test_image("astronaut")
camera.shape, camera.dtype, camera.min(), camera.max(), astronaut.shape
````

## Displaying images

**What we do.** We show both images side by side.

**What to expect.** The horizontal axis is the column index and the vertical axis is the row index of a pixel. Row 0 is at the **top**, as in a matrix, not at the bottom as in a usual graph.

**How to read the code.** `plt.subplots(1, 2, figsize=(10, 5))` creates a figure with one row of two panels (`axes[0]`, `axes[1]`); `figsize` is in inches. `images.show_image(image, title, ax=...)` draws an image into the given panel and labels its axes; `plt.tight_layout()` prevents overlapping labels and `plt.show()` displays the figure.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
images.show_image(camera, "camera (greyscale)", ax=axes[0])
images.show_image(astronaut, "astronaut (RGB)", ax=axes[1])
plt.tight_layout()
plt.show()
````

## Pixel values along one row

**What we do.** We plot the intensities of one row of the `camera` image as an ordinary graph.

**What to expect.** An image is just a function of two variables. Along a row, flat regions (sky) give a nearly constant curve, and the edges of objects (the photographer's coat) give sudden jumps between dark (small values) and bright (large values).

**How to read the code.** `camera[row, :]` selects all columns of one row (NumPy indexing is `[row, column]`). `plt.plot` draws the curve; `plt.xlabel` and `plt.ylabel` label the axes.

````python
row = 200
plt.figure(figsize=(10, 3))
plt.plot(camera[row, :])
plt.xlabel("column [pixel]")
plt.ylabel("intensity (0 = black, 255 = white)")
plt.title(f"row {row} of the camera image")
plt.show()
````

## RGB and YCbCr

**What we do.** We convert the colour image from RGB (red, green, blue) to YCbCr and show the three channels separately.

**What to expect.** `Y` (luminance) looks like a normal greyscale photograph and contains most of the detail. `Cb` and `Cr` (blue and red colour differences, centred at 128) look flat and blurry: the eye is much less sensitive to fine colour detail, which is why JPEG stores `Cb` and `Cr` at a lower resolution.

**How to read the code.** `images.rgb_to_ycbcr` returns an array of the same shape whose last axis holds `Y, Cb, Cr`. `ycbcr[:, :, index]` selects one channel. `enumerate` gives both the index and the name in the loop.

````python
ycbcr = images.rgb_to_ycbcr(astronaut)
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for index, name in enumerate(
    ["Y (luminance)", "Cb (blue difference)", "Cr (red difference)"]
):
    images.show_image(ycbcr[:, :, index], name, ax=axes[index])
plt.tight_layout()
plt.show()
````

## The conversion can be inverted

**What we do.** We convert back from YCbCr to RGB and measure the largest difference from the original image.

**What to expect.** A number of order `1e-13`: the conversion is a linear map with an invertible matrix, so the round trip is exact up to floating-point rounding.

**How to read the code.** `np.abs(a - b).max()` is the largest absolute difference between two arrays (the maximum norm of the error).

````python
np.abs(images.ycbcr_to_rgb(ycbcr) - astronaut).max()
````

## Fixed and scaled display

**What we do.** We darken the image by dividing all values by 4 and display it in two ways.

**What to expect.** With the fixed scale (0 = black, 255 = white) the darkened image looks dark, because its values are only in `[0, 64]`. With `scaled=True` the smallest value is shown as black and the largest as white (like Matlab's `imagesc`), so the image looks normal again: display scaling hides the true intensities. Use the scaled display for arrays that are not intensities, such as spectra or gradients.

**How to read the code.** `camera / 4` divides every pixel; the argument `scaled=True` switches the display mode.

````python
dark = camera / 4
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
images.show_image(dark, "fixed scale [0, 255]", ax=axes[0])
images.show_image(dark, "scaled to the data", ax=axes[1], scaled=True)
plt.tight_layout()
plt.show()
````

### Module `wavelets_lab.images` (file `src/wavelets_lab/images.py`)

Images as NumPy arrays and simple pixel operations. **Convention used everywhere in the course:** a greyscale image is a 2D array of type `float64` with values on the scale `[0, 255]` (0 = black, 255 = white); a colour image is a 3D array of shape `(rows, cols, 3)` on the same scale. Working in `float64` (instead of the `uint8` stored in image files) avoids overflow and rounding during computations.

The module starts with the imports
```python
from pathlib import Path
```

and the constant `MAX_INTENSITY = 255.0`.

#### `images.load_test_image(name="camera")`

**What it is.** Loads one of the licence-free test images shipped with scikit-image (`skimage.data`), so that no image file is needed.

**Input.**
- `name` (`str`, default `"camera"`): name of a function in `skimage.data`, e.g. `"camera"` (512×512 greyscale), `"astronaut"` (512×512 RGB), `"coins"`, `"checkerboard"`.

**Output.** `numpy.ndarray` of type `float64` on the scale `[0, 255]` (2D for greyscale, `(rows, cols, 3)` for colour). Raises `ValueError` if the name does not exist.

**How to write it yourself.**
1. Import the module of test images: `from skimage import data`.
2. Look the loader up by name without failing: `loader = getattr(data, name, None)`.
3. If `loader` is `None` or not callable (`callable(loader)` is `False`), raise `ValueError(f"skimage.data has no test image called {name!r}")`.
4. Call it: `raw = loader()` returns a `uint8` array.
5. Return `to_float(raw)` (see `images.to_float`), which converts to `float64` on `[0, 255]`.

#### `images.to_float(image)` (used internally)

**What it is.** Converts any image array to the course convention: `float64` on `[0, 255]`.

**Input.**
- `image` (array-like): `uint8`, `bool`, integer or floating-point image.

**Output.** `numpy.ndarray` of type `float64`.

**How to write it yourself.**
1. `array = np.asarray(image)`.
2. If `array.dtype == np.bool_`: return `array.astype(np.float64) * 255.0` (True becomes white).
3. If the dtype is floating (`np.issubdtype(array.dtype, np.floating)`), the array is not empty and all values lie in `[0, 1]` (`array.min() >= 0` and `array.max() <= 1`): the image uses the unit scale, so return `array.astype(np.float64) * 255.0`.
4. Otherwise (integers, or floats already on a large scale) return `array.astype(np.float64)`.

#### `images.show_image(image, title=None, ax=None, scaled=False, xlabel="column [pixel]", ylabel="row [pixel]")`

**What it is.** Displays a greyscale or colour image with Matplotlib, using a 256-level grey colour map and **labelled axes** (by default pixel coordinates: column on the horizontal axis, row on the vertical axis, with row 0 at the top).

**Input.**
- `image` (array-like): 2D greyscale or `(rows, cols, 3)` RGB image on `[0, 255]`.
- `title` (`str` or `None`): title of the plot.
- `ax` (`matplotlib.axes.Axes` or `None`): where to draw; when `None` a new figure is created. Passing `ax` lets you place several images in one figure (`fig, axes = plt.subplots(1, 3)`).
- `scaled` (`bool`, default `False`): `False` maps 0 to black and 255 to white (true intensities); `True` stretches the smallest value of the data to black and the largest to white, like Matlab's `imagesc`. Use `True` for arrays that are not intensities (spectra, gradients, phases).
- `xlabel`, `ylabel` (`str`): axis labels; change them when the axes mean something else (e.g. angle and offset for a sinogram).

**Output.** The `Axes` object that contains the image (useful for further changes such as `ax.set_xlim`).

**How to write it yourself.**
1. `array = np.asarray(image, dtype=np.float64)`.
2. If `ax is None`: `_, ax = plt.subplots()`.
3. If `array.ndim == 3` (colour): `ax.imshow(to_uint8(array))`; Matplotlib then uses the RGB values directly.
4. Otherwise (greyscale): choose limits `vmin, vmax = (None, None)` if `scaled` else `(0.0, 255.0)` and call `ax.imshow(array, cmap=plt.get_cmap("gray", 256), vmin=vmin, vmax=vmax)`. With `vmin=vmax=None` Matplotlib uses the data minimum and maximum.
5. If a title is given: `ax.set_title(title)`.
6. Label both axes: `ax.set_xlabel(xlabel)` and `ax.set_ylabel(ylabel)`.
7. Return `ax`.

#### `images.to_uint8(image)` (used internally)

**What it is.** The inverse direction: prepares an image for saving or for `imshow` of colour images, which expects bytes.

**Input.**
- `image` (array-like): image on the scale `[0, 255]`, possibly with values outside it.

**Output.** `numpy.ndarray` of type `uint8`.

**How to write it yourself.**
1. Convert to floats: `np.asarray(image, dtype=np.float64)`.
2. Round to the nearest integer with `np.rint`.
3. Clip to the valid range with `np.clip(..., 0, 255)`; without this, e.g. 260 would wrap around to 4.
4. Cast with `.astype(np.uint8)` and return.

#### `images.rgb_to_ycbcr(image)`

**What it is.** Converts an RGB image to the YCbCr colour space (standard ITU-R BT.601, used by JPEG). `Y` is the luminance (brightness, what a greyscale photo shows), `Cb` and `Cr` are blue and red colour differences centred at 128.

**Input.**
- `image` (array-like): RGB image of shape `(rows, cols, 3)` on `[0, 255]`.

**Output.** `numpy.ndarray` of the same shape with channels `Y, Cb, Cr`. Raises `ValueError` for a wrong shape.

**How to write it yourself.**
1. Define the conversion matrix (rows give `Y`, `Cb`, `Cr`):
   ```
   M = [[ 0.299,     0.587,     0.114   ],
        [-0.168736, -0.331264,  0.5     ],
        [ 0.5,      -0.418688, -0.081312]]
   ```
   and the offset `b = [0, 128, 128]`.
2. Check the input: convert with `np.asarray(image, dtype=np.float64)` and raise `ValueError` unless `ndim == 3` and the last dimension is 3 (internal helper `_check_colour`).
3. Every pixel is a row vector `p = (R, G, B)`; its YCbCr value is `M p + b`. For the whole image at once use matrix multiplication on the last axis: `rgb @ M.T + b`.
4. Return the result as `float64`.

#### `images.ycbcr_to_rgb(image)`

**What it is.** The inverse of `rgb_to_ycbcr`.

**Input.**
- `image` (array-like): YCbCr image of shape `(rows, cols, 3)`.

**Output.** RGB image of the same shape (not clipped, so tiny rounding errors may leave `[0, 255]`).

**How to write it yourself.**
1. Check the shape as in `rgb_to_ycbcr`.
2. Subtract the offset: `ycbcr - [0, 128, 128]`.
3. Multiply by the inverse matrix on the last axis: `(ycbcr - b) @ np.linalg.inv(M).T`.
4. Return as `float64`. A round trip `ycbcr_to_rgb(rgb_to_ycbcr(x))` reproduces `x` up to about `1e-12`.
