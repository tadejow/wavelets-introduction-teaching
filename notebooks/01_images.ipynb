# Week 1: an image as an array

Goals: load test images, inspect their type and range, convert colour images to YCbCr and display them.

```python
import matplotlib.pyplot as plt
import numpy as np

from wavelets_lab import images
```

## Loading images

Images are `float64` arrays on the scale [0, 255].

```python
camera = images.load_test_image("camera")
astronaut = images.load_test_image("astronaut")
camera.shape, camera.dtype, camera.min(), camera.max(), astronaut.shape
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
images.show_image(camera, "camera", ax=axes[0])
images.show_image(astronaut, "astronaut", ax=axes[1])
plt.show()
```

## RGB and YCbCr

The luminance Y carries most of the information; Cb and Cr are the colour differences.

```python
ycbcr = images.rgb_to_ycbcr(astronaut)
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for index, name in enumerate(["Y", "Cb", "Cr"]):
    images.show_image(ycbcr[:, :, index], name, ax=axes[index])
plt.show()
```

```python
np.abs(images.ycbcr_to_rgb(ycbcr) - astronaut).max()
```

## Scaled display

`scaled=True` stretches the grey levels to the data, like Matlab's `imagesc`.

```python
dark = camera / 4
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
images.show_image(dark, "fixed scale", ax=axes[0])
images.show_image(dark, "scaled", ax=axes[1], scaled=True)
plt.show()
```
